가상 과일 자료로 학습과 평가를 구분합니다. Python 표준 기능만 사용합니다.

셀을 위에서 아래로 실행하세요. 테스트 자료를 확인한 뒤 기준을 바꾸면 그 점수는 독립적인 최종 평가가 아닙니다.

## 1. 훈련 자료 준비
길이 단위는 cm입니다. A는 사과, B는 바나나이며, 수업용으로 만든 자료입니다.

```python
train = [(2,'A'),(3,'A'),(4,'A'),(5,'A'),(6,'B'),(7,'A'),(8,'B'),(9,'B')]

def predict(length, threshold):
    return 'A' if length < threshold else 'B'

def accuracy(data, threshold):
    return sum(predict(x, threshold) == y for x, y in data) / len(data)

print(train)
```

## 2. 훈련 자료로 경계 학습
후보 경계를 비교하여 훈련 정확도가 가장 높은 경계를 선택합니다. 동점이면 더 작은 경계를 사용합니다. 일반적인 개발에서는 별도 검증 자료로 모델 설정을 선택합니다.

```python
candidates = [2 + i * 0.5 for i in range(17)]
scores = [(t, accuracy(train, t)) for t in candidates]
best = max(scores, key=lambda pair: pair[1])[0]
for threshold, score in scores:
    print(f'경계 {threshold:4.1f}, 훈련 정확도 {score:.1%}')
print('선택한 경계:', best)
```

## 3. 선택을 마친 모델 평가
테스트 자료를 한 번 공개합니다. 경계를 변경하지 않고 평가하세요.

```python
test = [(3.5,'A'),(5.5,'B'),(7.5,'B'),(8.5,'B')]
for x, y in test:
    guess = predict(x, best)
    print(f'길이 {x}, 정답 {y}, 예측 {guess}, 일치 {guess == y}')
print(f'훈련 정확도: {accuracy(train, best):.1%}')
print(f'테스트 정확도: {accuracy(test, best):.1%}')
```

## 4. 경사하강법으로 가중치 조정
모형은 y = wx, 손실은 평균제곱오차입니다. 학습률을 0.05와 0.30으로 바꾸어 비교하세요. 이는 새 자료의 성능 평가가 아니라 최적화 과정의 체험입니다.

```python
samples = [(1,2),(2,4),(3,6)]
w = 0.0
learning_rate = 0.05
for step in range(11):
    loss = sum((w*x-y)**2 for x,y in samples) / len(samples)
    print(f'{step:2}회, w={w:.5f}, MSE={loss:.5f}')
    gradient = 2 * sum(x*(w*x-y) for x,y in samples) / len(samples)
    w -= learning_rate * gradient
```
