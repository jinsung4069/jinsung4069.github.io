### B-2. 데이터 불러오기
기본값은 펭귄 자료입니다. 자신의 CSV 파일을 쓰려면 `USE_MY_FILE = True` 로 바꾸고 실행한 뒤 파일을 고릅니다. 개인정보가 들어 있는 파일은 올리지 않습니다.

### B-3. 모델의 종류와 열 고르기
`TASK` 에는 `분류`, `예측`, `군집` 중 하나를 적습니다. 군집이면 `TARGET` 은 `None` 으로 둡니다. 특징 열은 숫자로 된 열을 고릅니다.

```python
TASK = '분류'
FEATURES = ['bill_length_mm', 'flipper_length_mm']
TARGET = 'species'
K = 5            # 분류에서 참고할 이웃의 수, 군집에서 만들 군집의 수
TEST_SIZE = 0.25

cols = FEATURES + ([TARGET] if TARGET else [])
clean = data.dropna(subset=cols).reset_index(drop=True)
print(f'빈 값이 있는 행을 지운 뒤 {len(clean)}행 (처음 {len(data)}행)')
```

### B-4. 학습과 평가

```python
X = clean[FEATURES]
if TASK == '분류':
    y = clean[TARGET]
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=TEST_SIZE, stratify=y, random_state=SEED)
    model = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=K)).fit(X_train, y_train)
    pred = model.predict(X_test)
    labels = sorted(y.unique())
    print(f'평가 자료 정확도 {accuracy_score(y_test, pred):.1%} ({(pred == y_test).sum()} / {len(y_test)})')
    display(pd.DataFrame(confusion_matrix(y_test, pred, labels=labels), index=[f'실제 {l}' for l in labels], columns=[f'예측 {l}' for l in labels]))
elif TASK == '예측':
    y = clean[TARGET]
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=TEST_SIZE, random_state=SEED)
    model = LinearRegression().fit(X_train, y_train)
    pred = model.predict(X_test)
    print(f'결정계수 R² {r2_score(y_test, pred):.2f}, 평균 절대 오차 {mean_absolute_error(y_test, pred):.2f}')
    for name, coef in zip(FEATURES, model.coef_):
        print(f'{name} 이 1 늘면 {TARGET} 이 {coef:.2f} 변합니다.')
elif TASK == '군집':
    model = make_pipeline(StandardScaler(), KMeans(n_clusters=K, n_init=10, random_state=SEED)).fit(X)
    clean['cluster'] = model[-1].labels_
    display(clean.groupby('cluster')[FEATURES].mean().assign(count=clean['cluster'].value_counts().sort_index()))
```

### B-5. 새 입력으로 확인하기
학습과 평가에 쓰지 않은 새 값을 넣어 프로그램처럼 사용해 봅니다. 특징 열의 순서에 맞게 값을 적습니다.

```python
new_values = [[45.0, 200.0]]
new = pd.DataFrame(new_values, columns=FEATURES)
result = model.predict(new)
print('입력', new_values[0], '결과', result[0])
if TASK == '분류':
    print('이웃 비율', dict(zip(model.classes_, model.predict_proba(new)[0].round(2))))
```
